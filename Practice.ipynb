```python
largest = None
print('Before:', largest)
for itervar in [3, 41, 12, 9, 74, 15]:
  if largest is None or itervar > largest :
    largest = itervar
print('Loop:', itervar, largest)
print('Largest:', largest)
```

```python
smallest = None
print('Before:', smallest)
for itervar in [3, 41, 12, 9, 74, 15]:
  if smallest is None or itervar < smallest:
    smallest = itervar
print('Loop:', itervar, smallest)
print('Smallest:', smallest)
```

```python
# Initialize variables
total = 0
count = 0

# Keep asking for numbers until the user enters "done"
while True:
    user_input = input("Enter an integer or type 'done': ")

    # Stop the loop when the user enters "done"
    if user_input.lower() == "done":
        break

    # Try to convert the input into an integer
    try:
        number = int(user_input)

        # Add the number to the total and increase the count
        total += number
        count += 1

    # Handle inputs that are not integers
    except ValueError:
        print("Error: Please enter a valid integer.")
        continue

# Calculate and display the results
if count > 0:
    average = total / count
    print("Total:", total)
    print("Count:", count)
    print("Average:", average)
else:
    print("No integers were entered.")
```

```python
# Ask the user to enter the name of the file
filename = input("Enter a file name: ")

# Open the file for reading
try:
    file = open(filename, "r")

    # Read and print each line in uppercase
    for line in file:
        print(line.upper(), end="")

    # Close the file
    file.close()

except FileNotFoundError:
    print("File not found. Please check the file name.")

```

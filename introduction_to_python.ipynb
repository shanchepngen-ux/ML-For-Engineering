```python
import pandas as pd

file_path = '/content/drive/My Drive/Colab Notebooks/AmazonReviews - AmazonReviews.csv'

try:
    df = pd.read_csv(file_path)
    print(f"Successfully loaded '{file_path}' into DataFrame.")
    display(df.head())
except FileNotFoundError:
    print(f"Error: '{file_path}' not found. Please ensure the file is in the correct location and the name is exact.")
except Exception as e:
    print(f"An error occurred: {e}")
```

### 1. Get Basic Information about the DataFrame

This will show us the number of entries, columns, data types, and non-null values, which is crucial for identifying potential issues like missing data or incorrect data types.

```python
print(df.info())
```

### 2. Display Descriptive Statistics

For numerical columns, this provides summary statistics like count, mean, standard deviation, min, max, and quartiles. This helps in understanding the distribution of numerical data.

```python
display(df.describe(include='all'))
```

### 3. Check for Missing Values

It's important to know if there are any missing values in your dataset and in which columns. This helps in deciding how to handle them (e.g., filling, dropping).

```python
print('Missing values per column:')
print(df.isnull().sum())
```

### 4. Drop the 'variant:' Column

Since the `variant:` column has 338 missing values (all of its entries), it provides no useful information. We can safely drop it from the DataFrame.

```python
df = df.drop(columns=['variant:'])
print("Dropped 'variant:' column. Current DataFrame columns:")
print(df.columns)
```

### 5. Convert 'rating' Column to Numerical Type

The `rating` column is currently a string (e.g., '5.0 out of 5 stars'). To use it for numerical analysis, we need to extract the numerical part (e.g., '5.0') and convert it to a float.

```python
df['rating'] = df['rating'].apply(lambda x: float(x.split(' ')[0]))
print("Converted 'rating' column to float. Displaying info and first few rows:")
print(df.info())
display(df.head())
```
